# NumPy — a estrutura de dados do Aprendizado de Máquina

Em Aprendizado de Máquina, um conjunto de dados é quase sempre representado por:

- uma **matriz de atributos** $X$, de dimensão $m \times n$, em que
  - $m$ é o número de **exemplos** (linhas, ou objetos observados);
  - $n$ é o número de **atributos** (colunas, ou variáveis medidas);
- um **vetor de rótulos** $y$, com $m$ elementos.

$$
X =
\begin{bmatrix}
x_1^{(1)} & x_2^{(1)} & \cdots & x_n^{(1)} \\
x_1^{(2)} & x_2^{(2)} & \cdots & x_n^{(2)} \\
\vdots & \vdots & \ddots & \vdots \\
x_1^{(m)} & x_2^{(m)} & \cdots & x_n^{(m)}
\end{bmatrix}
\qquad
y =
\begin{bmatrix}
y^{(1)} \\ y^{(2)} \\ \vdots \\ y^{(m)}
\end{bmatrix}
$$

O NumPy é a biblioteca que nos dá essa estrutura: o **`ndarray`**.

```python
import numpy as np
```

## Criando arrays

```python
# A partir de uma lista
v = np.array([2.0, 4.0, 6.0, 8.0])

# A partir de uma lista de listas -> matriz
X = np.array([
    [1.0, 1.0],
    [1.5, 1.8],
    [2.0, 1.0],
    [4.0, 4.0],
    [4.5, 5.0],
    [5.0, 4.2]
])

y = np.array([0, 0, 0, 1, 1, 1])

print('v =', v)
print()
print('X =')
print(X)
print()
print('y =', y)
```

<p align=justify>
Todo <code>array</code> possui atributos que descrevem sua estrutura. O mais importante para
nós é o <code>shape</code>, que informa quantas linhas e quantas colunas ele possui.
</p>

```python
print('Formato de X (linhas, colunas):', X.shape)
print('Número de exemplos (m):', X.shape[0])
print('Número de atributos (n):', X.shape[1])
print('Número de dimensões:', X.ndim)
print('Tipo dos elementos:', X.dtype)
print()
print('Formato de y:', y.shape)
```

<p align=justify>
Repare que <code>y.shape</code> retorna <code>(6,)</code>, e não <code>(6, 1)</code>. Trata-se
de um <b>vetor unidimensional</b>. Essa distinção causa muitos erros no início — sempre que
uma operação falhar, verifique os <code>shape</code> envolvidos.
</p>

## Indexação e fatiamento

A indexação de um `array` bidimensional segue o padrão `X[linha, coluna]`. Os índices
começam em `0` e o símbolo `:` significa "todos".

```python
print('X completo:'); print(X)
print()
print('Elemento da linha 0, coluna 1:', X[0, 1])
print()
print('Primeira linha (exemplo 0):', X[0])
print()
print('Primeira coluna (atributo x1):', X[:, 0])
print()
print('Três primeiras linhas:'); print(X[:3])
print()
print('Últimas duas linhas, apenas a coluna 1:', X[-2:, 1])
```

### Máscaras booleanas

Este é, provavelmente, o recurso mais usado do NumPy nesta disciplina: selecionar elementos
a partir de uma **condição**.

```python
mascara = (y == 0)

print('Máscara booleana:', mascara)
print()
print('Exemplos da classe 0:')
print(X[mascara])
print()
print('Exemplos da classe 1:')
print(X[y == 1])
```

```python
# Condições também podem ser combinadas com & (e), | (ou), ~ (não)
condicao = (X[:, 0] > 1.5) & (X[:, 1] < 4.5)

print('Exemplos com x1 > 1.5 E x2 < 4.5:')
print(X[condicao])
```

<p align=justify>
Cuidado: dentro do NumPy usamos <code>&</code>, <code>|</code> e <code>~</code> em vez de
<code>and</code>, <code>or</code> e <code>not</code>. Além disso, cada condição precisa estar
entre parênteses.

## Vetorização

Operações aritméticas em `arrays` são aplicadas **elemento a elemento**, sem necessidade de
laços. Isso deixa o código mais curto, mais legível e muito mais rápido.

```python
a = np.array([1.0, 2.0, 3.0, 4.0])
b = np.array([10.0, 20.0, 30.0, 40.0])

print('a + b =', a + b)
print('a * b =', a * b)
print('b / a =', b / a)
print('a ** 2 =', a ** 2)
print('2 * a + 1 =', 2 * a + 1)
print()
print('Raiz quadrada:', np.sqrt(b))
print('Exponencial:', np.exp(a))
print('Logaritmo natural:', np.log(b))
```

### Por que isso importa?

Vamos comparar o tempo de execução de um laço `for` com o de uma operação vetorizada.

```python
import time

n = 1_000_000
x = rng.random(n)

# Versão com laço
inicio = time.time()
soma_quadrados = 0.0
for valor in x:
    soma_quadrados += valor ** 2
tempo_laco = time.time() - inicio

# Versão vetorizada
inicio = time.time()
soma_quadrados_np = np.sum(x ** 2)
tempo_numpy = time.time() - inicio

print(f'Laço for  : {tempo_laco:.4f} s')
print(f'NumPy     : {tempo_numpy:.4f} s')
print(f'Aceleração: {tempo_laco / tempo_numpy:.1f}x')
```

<p align=justify>
A diferença costuma ser de uma a duas ordens de grandeza. Quando trabalharmos com milhares de
exemplos e dezenas de atributos, escrever código vetorizado deixa de ser elegância e passa a
ser necessidade.
</p>

## Reduções e o conceito de eixo (`axis`)

Funções como `sum`, `mean`, `std`, `min` e `max` podem ser aplicadas ao array inteiro ou ao
longo de um **eixo**:

- `axis=0` → percorre as **linhas**, produzindo um resultado **por coluna** (por atributo);
- `axis=1` → percorre as **colunas**, produzindo um resultado **por linha** (por exemplo).

Na prática, `axis=0` é o que usamos quase sempre, porque queremos estatísticas **por
atributo**.

```python
print('X:'); print(X)
print()
print('Média geral (todos os valores):', X.mean())
print('Média por coluna  (axis=0):', X.mean(axis=0))
print('Média por linha   (axis=1):', X.mean(axis=1))
print()
print('Desvio padrão por coluna:', X.std(axis=0))
print('Mínimo por coluna:', X.min(axis=0))
print('Máximo por coluna:', X.max(axis=0))
```

```python
# argmin / argmax devolvem a POSIÇÃO do menor/maior valor, e não o valor em si.
# Serão usados na próxima aula, para encontrar o vizinho mais próximo.

distancias = np.array([3.2, 0.8, 5.1, 1.4])

print('Menor distância:', distancias.min())
print('Índice da menor distância:', np.argmin(distancias))
print('Ordem crescente dos índices:', np.argsort(distancias))
```

## Broadcasting

O *broadcasting* permite combinar arrays de formatos diferentes, desde que compatíveis. 

O NumPy "estica" automaticamente o array menor.

Um uso importante é na **padronização** dos atributos:

$$
z^{(i)}_j = \frac{x^{(i)}_j - \mu_j}{\sigma_j}
$$

em que $\mu_j$ e $\sigma_j$ são a média e o desvio padrão do atributo $j$.

```python
medias = X.mean(axis=0)     # shape (2,)
desvios = X.std(axis=0)     # shape (2,)

print('X.shape      :', X.shape)
print('medias.shape :', medias.shape)
print()

X_padronizado = (X - medias) / desvios   # (6,2) combinado com (2,) -> broadcasting

print('X padronizado:')
print(np.round(X_padronizado, 3))
print()
print('Média após padronização (deve ser ~0):', np.round(X_padronizado.mean(axis=0), 10))
print('Desvio após padronização (deve ser 1):', np.round(X_padronizado.std(axis=0), 10))
```

<p align=justify>
Guarde essa ideia: a padronização será decisiva na aula sobre k-NN, porque métodos baseados
em distância são sensíveis à escala dos atributos.
</p>

## Um pouco de álgebra linear

Modelos lineares, tais como regressão linear, regressão logística, perceptron, são escritos como
produtos entre matrizes e vetores. 

O NumPy oferece o operador `@` para isso.

```python
w = np.array([0.5, -1.2])   # vetor de pesos (um por atributo)
b = 0.3                     # termo de viés (intercepto)

# Produto interno entre dois vetores
print('Produto interno x^(0) . w =', X[0] @ w)
print()

# Produto matriz-vetor: calcula a saída para TODOS os exemplos de uma só vez
saidas = X @ w + b
print('Saída do modelo para cada exemplo:')
print(np.round(saidas, 3))
```

```python
A = np.array([[1.0, 2.0],
              [3.0, 4.0]])

print('Transposta de A:'); print(A.T)
print()
print('Produto A @ A:'); print(A @ A)
print()
print('Inversa de A:'); print(np.round(np.linalg.inv(A), 3))
print()
print('Norma (comprimento) do vetor w:', np.linalg.norm(w))
```

```python
import pandas as pd
```

## Selecionando colunas e linhas

Existem duas formas principais de seleção:

- **`.loc[linhas, colunas]`** → seleciona por **rótulo** (nome);
- **`.iloc[linhas, colunas]`** → seleciona por **posição** (número inteiro).

```python
# Uma coluna -> Series
df['prova1']
```

```python
# Várias colunas -> DataFrame (note a lista dentro dos colchetes)
df[['aluno', 'prova1', 'prova2']]
```

```python
print('--- .loc (por rótulo) ---')
print(df.loc[0, 'aluno'])
print()
print(df.loc[0:2, ['aluno', 'prova1']])
print()
print('--- .iloc (por posição) ---')
print(df.iloc[0, 0])
print()
print(df.iloc[0:3, [0, 3]])
```

<p align=justify>
Atenção a uma diferença sutil: em <code>.loc</code>, o intervalo <code>0:2</code>
<b>inclui</b> a linha 2; em <code>.iloc</code>, <code>0:3</code> <b>exclui</b> a posição 3,
como é o padrão do Python.
</p>

## Filtrando linhas

A lógica é a mesma das máscaras booleanas do NumPy.

```python
df[df['faltas'] <= 5]
```

```python
df[(df['curso'] == 'EP') & (df['prova2'] >= 6.0)]
```

## Criando e transformando colunas

```python
df['media'] = 0.4 * df['prova1'] + 0.6 * df['prova2']
df['aprovado'] = (df['media'] >= 6.0) & (df['faltas'] <= 10)

df
```

```python
# apply aplica uma função a cada elemento da coluna
def conceito(nota):
    if nota >= 9.0:
        return 'A'
    elif nota >= 7.0:
        return 'B'
    elif nota >= 6.0:
        return 'C'
    return 'D'


df['conceito'] = df['media'].apply(conceito)
df[['aluno', 'media', 'conceito']]
```

```python
# groupby: divide -> aplica -> combina
df.groupby('curso')[['prova1', 'prova2', 'media', 'faltas']].mean()
```

```python
# Várias estatísticas de uma vez
df.groupby('curso')['media'].agg(['count', 'mean', 'std', 'min', 'max'])
```

<p align=justify>
O <code>groupby</code> será muito útil quando quisermos comparar o comportamento dos atributos

</p>

## Valores faltantes

Conjuntos de dados reais quase sempre possuem valores ausentes (`NaN`). Precisamos ao menos
saber detectá-los e tratá-los.

```python
df_faltante = df.copy()
df_faltante.loc[1, 'prova1'] = np.nan
df_faltante.loc[4, 'faltas'] = np.nan

print('Valores faltantes por coluna:')
print(df_faltante.isna().sum())
```

```python
# Opção 1: remover as linhas com faltantes
print('Linhas antes:', len(df_faltante))
print('Linhas depois de dropna():', len(df_faltante.dropna()))
print()

# Opção 2: preencher com alguma estatística (imputação)
df_preenchido = df_faltante.fillna({
    'prova1': df_faltante['prova1'].mean(),
    'faltas': df_faltante['faltas'].median()
})

print('Faltantes após o preenchimento:', df_preenchido.isna().sum().sum())
```

<p align=justify>
Não existe uma resposta única para o tratamento de valores faltantes. Descartar linhas é
simples, mas pode jogar fora muita informação; imputar pela média preserva o tamanho da
amostra, mas reduz artificialmente a variabilidade dos dados. A escolha depende do problema.
</p>

## Gráfico de dispersão

É o gráfico mais usado na disciplina: cada ponto é um exemplo, cada eixo é um atributo, e a
cor indica a classe.

# Juntando tudo: análise exploratória do conjunto Iris

Vamos aplicar as três bibliotecas a um conjunto de dados clássico. O **Iris** contém 150
flores de três espécies, descritas por quatro atributos (em centímetros):

- comprimento e largura da **sépala**;
- comprimento e largura da **pétala**.

O objetivo, nas próximas aulas, será prever a espécie a partir desses quatro atributos.

```python
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)

dados = iris.frame                       # DataFrame com atributos + coluna 'target'
dados['especie'] = dados['target'].map(dict(enumerate(iris.target_names)))

dados.head()
```

## Inspeção inicial

```python
print('Formato:', dados.shape)
print()
print('Valores faltantes por coluna:')
print(dados.isna().sum())
print()
print('Exemplos por espécie:')
print(dados['especie'].value_counts())
```

```python
dados.describe().round(2)
```

## Comparando as espécies

```python
atributos = iris.feature_names

dados.groupby('especie')[atributos].mean().round(2)
```

<p align=justify>
Repare que os atributos da <b>pétala</b> variam muito mais entre as espécies do que os da
sépala. Isso é um primeiro indício de que a pétala carrega mais informação para separar as
classes — hipótese que o gráfico a seguir confirma.
</p>

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for especie in dados['especie'].unique():
    subconjunto = dados[dados['especie'] == especie]

    axes[0].scatter(subconjunto[atributos[0]], subconjunto[atributos[1]],
                    s=45, alpha=0.8, label=especie)
    axes[1].scatter(subconjunto[atributos[2]], subconjunto[atributos[3]],
                    s=45, alpha=0.8, label=especie)

axes[0].set_xlabel('Comprimento da sépala (cm)')
axes[0].set_ylabel('Largura da sépala (cm)')
axes[0].set_title('Sépala')

axes[1].set_xlabel('Comprimento da pétala (cm)')
axes[1].set_ylabel('Largura da pétala (cm)')
axes[1].set_title('Pétala')

for ax in axes:
    ax.legend()
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

### Leitura dos gráficos

No espaço das **sépalas**, as espécies *versicolor* e *virginica* se sobrepõem bastante.
No espaço das **pétalas**, as três espécies formam grupos visivelmente mais separados.

Essa observação antecipa uma ideia central da disciplina:

> a dificuldade de um problema de classificação depende fortemente dos **atributos**
> escolhidos para representar os objetos.

```python
# Distribuição de cada atributo, por espécie
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for ax, atributo in zip(axes.ravel(), atributos):
    for especie in dados['especie'].unique():
        ax.hist(dados[dados['especie'] == especie][atributo],
                bins=15, alpha=0.6, label=especie)
    ax.set_title(atributo)
    ax.set_xlabel('cm')
    ax.set_ylabel('frequência')
    ax.legend(fontsize=8)

plt.tight_layout()
plt.show()
```

```python
# Correlação entre os atributos
dados[atributos].corr().round(2)
```

<p align=justify>
Comprimento e largura da pétala são fortemente correlacionados. Atributos redundantes não
necessariamente prejudicam um modelo, mas motivam o tema da aula sobre <b>redução de
dimensionalidade</b>, no fim do curso.
</p>
