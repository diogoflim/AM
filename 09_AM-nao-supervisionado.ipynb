```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
```

```python
# Geração de dados em 3 clusters
X, y = make_blobs(n_samples=150,
                  centers=3,
                  cluster_std=0.6,
                  random_state=0)

# Visualização
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis', edgecolor='black', s=60)
plt.title('Exemplo de dados com 3 clusters')
plt.xlabel('Atributo 1')
plt.ylabel('Atributo 2')
plt.grid(True)
plt.show()
```

```python
from sklearn.cluster import KMeans
import numpy as np
```

```python

# Criar dados
X, _ = make_blobs(n_samples=150, centers=3, cluster_std=0.60, random_state=42)

# Passo 1: centróides aleatórios
kmeans_random = KMeans(n_clusters=3, init='random', n_init=1, max_iter=1, random_state=42)
kmeans_random.fit(X)
centroids_init = kmeans_random.cluster_centers_

# Visualizar os dados + centróides iniciais
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], s=50, c='lightgray', edgecolor='black')
plt.scatter(centroids_init[:, 0], centroids_init[:, 1], s=200, c='red', marker='X', label='Centróides iniciais')
plt.title('Passo 1: Inicialização dos centróides')
plt.xlabel('Atributo 1')
plt.ylabel('Atributo 2')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
